# Exploration Notebook

This notebook is intended for exploratory data analysis and experimentation with the AIWordingAssist application. It will cover various aspects such as data ingestion, processing, and model evaluation.

```python
# Import necessary libraries
import os
import pandas as pd
from app.ingestion.pdf_loader import PDFLoader
from app.services.rag_service import RAGService
from app.evaluation.ragas_evaluator import RAGEvaluator

# Set up paths
raw_data_path = os.path.join('..', 'data', 'raw')
processed_data_path = os.path.join('..', 'data', 'processed_chunks')

# Load and explore raw documents
def load_raw_documents(path):
    documents = []
    for filename in os.listdir(path):
        if filename.endswith('.pdf'):
            loader = PDFLoader(os.path.join(path, filename))
            documents.append(loader.load())
    return documents

raw_documents = load_raw_documents(raw_data_path)
print(f'Loaded {len(raw_documents)} documents.')

# Example of processing documents with RAGService
rag_service = RAGService()
processed_docs = rag_service.process_documents(raw_documents)
print(f'Processed {len(processed_docs)} documents.')

# Evaluation of the processed documents
evaluator = RAGEvaluator()
evaluation_results = evaluator.evaluate(processed_docs)
print('Evaluation Results:', evaluation_results)
```
