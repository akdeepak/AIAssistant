# Evaluation Notebook

This notebook is designed to evaluate the performance of the AIWordingAssist application. It will cover various metrics and methods to assess the effectiveness of the retrieval and generation processes.

```python
# Import necessary libraries
import pandas as pd
import numpy as np
from app.evaluation.ragas_evaluator import RAGEvaluator
from app.evaluation.retrieval_metrics import RetrievalMetrics

# Initialize evaluator and metrics
rag_evaluator = RAGEvaluator()
retrieval_metrics = RetrievalMetrics()

# Load evaluation datasets
def load_evaluation_data(path):
    return pd.read_csv(path)

# Example usage
evaluation_data = load_evaluation_data('data/evaluation_sets/sample_evaluation_data.csv')

# Evaluate the model
def evaluate_model(data):
    results = rag_evaluator.evaluate(data)
    metrics = retrieval_metrics.calculate_metrics(data)
    return results, metrics

# Run evaluation
results, metrics = evaluate_model(evaluation_data)

# Display results
print("Evaluation Results:", results)
print("Retrieval Metrics:", metrics)
```
